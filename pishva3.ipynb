```python
import gradio as gr
import requests

def summarize(text, key):
    if not key: return " API Key لازم است"
    if len(text.strip()) < 30: return " متن خیلی کوتاه است"
    r = requests.post(
        "https://openrouter.ai/api/v1/chat/completions",
        headers={"Authorization": f"Bearer {key}", "Content-Type": "application/json"},
        json={
            "model": "openai/gpt-3.5-turbo",
            "messages": [{"role": "system", "content": "تو یک خلاصه‌ساز هستی."},
                         {"role": "user", "content": f"خلاصه کن:\n{text}"}]
        }
    )
    return r.json()["choices"][0]["message"]["content"]

gr.Interface(
    fn=summarize,
    inputs=[
        gr.Textbox(label=" متن", lines=4),
        gr.Textbox(label=" API Key", type="password")
    ],
    outputs=gr.Textbox(label=" خلاصه", lines=4),
    title=" خلاصه‌ساز متن"
).launch()
```
